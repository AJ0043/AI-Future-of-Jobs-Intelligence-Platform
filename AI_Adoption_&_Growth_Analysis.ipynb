```python
#Q3 — Current vs Projected Openings by Job Title
job_openings = (
    ai_data.groupby('Job Title')[
        ['Job Openings (2024)', 'Projected Openings (2030)']
    ]
    .sum()
)

job_openings.head()
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(24, 8))


# =========================================================
# Q1 — BAR CHART
# =========================================================

sample_20 = EDA_ai_data.sample(10)

bars = axes[0].bar(
    sample_20['Industry'],
    sample_20['Job Openings (2024)']
)

axes[0].set_title(
    'Q1 — 2024 Job Openings',
    fontsize=13,
    fontweight='bold'
)

axes[0].set_xlabel(
    'Industry',
    fontsize=11,
    fontweight='bold'
)

axes[0].set_ylabel(
    'Job Openings',
    fontsize=11,
    fontweight='bold'
)

axes[0].tick_params(
    axis='x',
    rotation=45
)

# Value labels on bars
axes[0].bar_label(
    bars,
    labels=[
        f'{int(value):,}'
        for value in sample_20['Job Openings (2024)']
    ],
    padding=3,
    fontsize=9
)


# =========================================================
# Q2 — DONUT CHART
# =========================================================

sample_20 = project_opening_2024.sample(5)

wedges, texts, autotexts = axes[1].pie(
    sample_20.values,
    labels=None,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops=dict(width=0.4),
    pctdistance=0.78
)

axes[1].set_title(
    'Q2 — 2030 Projected Openings',
    fontsize=13,
    fontweight='bold'
)

# Donut center label
axes[1].text(
    0,
    0,
    '2030\nProjected',
    ha='center',
    va='center',
    fontsize=11,
    fontweight='bold'
)

# Legend with Industry + value
legend_labels = [
    f'{industry} — {int(value):,}'
    for industry, value
    in zip(sample_20.index, sample_20.values)
]

axes[1].legend(
    wedges,
    legend_labels,
    title='Industry',
    loc='lower center',
    bbox_to_anchor=(0.5, -0.25),
    fontsize=8
)


# =========================================================
# Q3 — SCATTER PLOT
# =========================================================

sample_scatter = EDA_ai_data.sample(20)

x = sample_scatter['Job Openings (2024)']
y = sample_scatter['Projected Openings (2030)']

scatter = axes[2].scatter(
    x,
    y,
    s=70
)

axes[2].set_title(
    'Q3 — Current vs Projected Openings',
    fontsize=13,
    fontweight='bold'
)

axes[2].set_xlabel(
    '2024 Job Openings',
    fontsize=11,
    fontweight='bold'
)

axes[2].set_ylabel(
    '2030 Projected Openings',
    fontsize=11,
    fontweight='bold'
)

# Label every point
for i in range(len(sample_scatter)):

    industry = sample_scatter['Industry'].iloc[i]
    x_value = x.iloc[i]
    y_value = y.iloc[i]

    axes[2].annotate(
        f'{industry}\n({int(x_value):,}, {int(y_value):,})',
        (x_value, y_value),
        fontsize=7,
        xytext=(5, 5),
        textcoords='offset points'
    )


# =========================================================
# MAIN TITLE
# =========================================================

plt.suptitle(
    'Job Market Analysis — Q1, Q2 & Q3',
    fontsize=17,
    fontweight='bold'
)

plt.tight_layout()

plt.show()
```

```python
#Q5. Which industries and job roles offer the highest median salaries?
industry_salary = (
    EDA_ai_data
    .groupby('Industry')['Median Salary (USD)']
    .median()
    .sort_values(ascending=False)
)
```

```python
industry_salary=industry_salary.sample(5)
```

```python
#Q6. How does median salary vary with required experience?
salary_experience = (
    EDA_ai_data
    .groupby('Experience Required (Years)')['Median Salary (USD)']
    .median()
    .sort_index()
)

salary_experience
```

```python
#Q7. Does higher education requirement correspond to higher median salary?
education_salary = (
    EDA_ai_data
    .groupby('Required Education')['Median Salary (USD)']
    .median()
    .sort_values(ascending=False)
)

education_salary
```

```python
fig, axes = plt.subplots(1, 3, figsize=(24, 8))

# =============================================
# Bar Chart
# =============================================

bars = axes[0].bar(
    industry_salary.index,
    industry_salary.values
)

axes[0].set_title(
    'Q5 — Median Salary by Industry',
    fontsize=13,
    fontweight='bold'
)

axes[0].set_xlabel('Industry')
axes[0].set_ylabel('Median Salary (USD)')

axes[0].tick_params(
    axis='x',
    rotation=45
)

# Salary labels
axes[0].bar_label(
    bars,
    labels=[f'${x:,.0f}' for x in industry_salary.values],
    padding=2,
    fontsize=8
)

# =============================================
# Pie Chart
# =============================================
wedges, texts, autotexts = axes[1].pie(
    salary_experience.values,
    labels=None,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops=dict(width=0.4),
    pctdistance=0.76,
    textprops=dict(fontsize=9),
    colors=sns.color_palette('pastel')
)

axes[1].set_title(
    'Q6 — Salary Distribution by Experience',
    fontsize=13,
    fontweight='bold'
)

axes[1].legend(
    wedges,
    salary_experience.index,
    title='Experience (Years)',
    loc='center left',
    bbox_to_anchor=(1, 0.5)
)


# =============================================
# Q7 — Line Chart
# =============================================

axes[2].plot(
    education_salary.index,
    education_salary.values,
    marker='o',
    linewidth=2
)

axes[2].set_title(
    'Q7 — Median Salary by Required Education',
    fontsize=13,
    fontweight='bold'
)

axes[2].set_xlabel('Required Education')
axes[2].set_ylabel('Median Salary (USD)')

axes[2].tick_params(
    axis='x',
    rotation=45
)

axes[2].grid(
    axis='y',
    linestyle='--',
    alpha=0.4
)

plt.tight_layout()







plt.tight_layout()
plt.show()
```

```python
# How does AI Impact level vary  across industries ?
ai_impact_level = (
    EDA_ai_data.groupby('Industry')['AI Impact Level']
    .value_counts(normalize= True)
    .sort_index()
    .unstack()

)
```

```python
ai_impact_level
```

***Q. How does AI Impact level vary  across industries ?

 Ans:- The analysis shows that the average Automation Risk is relatively similar across High, Low, and Moderate AI Impact Levels for all industries. The values remain close to approximately 0.32–0.35, indicating that there is no strong variation in Automation Risk based solely on AI Impact Level.

Among the industries, Entertainment shows the highest value for Moderate AI Impact (0.3512), while Finance has relatively similar values across all three AI Impact Levels. Overall, the results suggest that AI Impact Level alone does not create a substantial difference in Automation Risk across industries

```python
#Q9. What is the relationship between AI Impact Level and Automation Risk?
relation_data = (
    EDA_ai_data
    .groupby('AI Impact Level')['Automation Risk (%)']
    .mean()
    .sort_values(ascending=False)
)

relation_data
```

Q9. What is the relationship between AI Impact Level and Automation Risk?

Ans:- The analysis shows that Automation Risk is relatively similar across different AI Impact Levels. The average automation risk does not show a strong difference between High, Moderate, and Low AI Impact Levels.

This suggests that AI Impact Level alone is not a strong indicator of Automation Risk in this dataset. Jobs with different levels of AI impact can have relatively similar automation risks.

Conclusion: There is no clear or strong relationship between AI Impact Level and Automation Risk based on the average values in this dataset. Other factors such as industry, job role, experience, and education requirements may also influence automation risk.
