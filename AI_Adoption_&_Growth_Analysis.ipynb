Q11. Which job roles show the largest change in openings between 2024 and 2030?

The analysis compares the number of job openings in 2024 with the projected openings in 2030 for each job role. The change is calculated as:

Change in Openings = Projected Openings (2030) − Job Openings (2024)

Job roles with a higher positive change are expected to experience a greater increase in openings by 2030, while negative values indicate a projected decline.

Conclusion: The job roles with the largest positive changes show the strongest projected growth in demand between 2024 and 2030. This comparison helps identify the roles with the greatest expected change in job opportunities.

```python
# ============================================================
# Q11 — Largest Change in Job Openings (2024 vs 2030)
# ============================================================

q11_data = (
    EDA_ai_data
    .groupby('Job Title')[
        ['Job Openings (2024)', 'Projected Openings (2030)']
    ]
    .sum()
    .reset_index()
)

# Calculate change in openings
q11_data['Opening Change'] = (
    q11_data['Projected Openings (2030)']
    - q11_data['Job Openings (2024)']
)

# Sort by largest change
q11_data = q11_data.sort_values(
    by='Opening Change',
    ascending=False
)

# Top 10 job roles
q11_top10 = q11_data.head(10)

q11_top10
```

```python
# Q12 — How does projected job growth vary across industries?

q12_data = (
    EDA_ai_data
    .groupby("Industry")[[
        "Job Openings (2024)",
        "Projected Openings (2030)"
    ]]
    .sum()
)

q12_data
```

Q12. How does projected job growth vary across industries?

The analysis compares job openings in 2024 with projected openings in 2030 across different industries. Projected job growth is calculated by subtracting the 2024 openings from the projected 2030 openings.

# **Feature Extraction**

```python
# Select numerical features for Feature Extraction

numerical_features = [
    "Median Salary (USD)",
    "Experience Required (Years)",
    "Job Openings (2024)",
    "Projected Openings (2030)",
    "Remote Work Ratio (%)",
    "Automation Risk (%)",
    "Gender Diversity (%)",
    "Opening_Change",
    "Growth_%"
]

X = Featuring[numerical_features]
```

```python
# Standardize numerical features before PCA

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
# Apply PCA for Feature Extraction

from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)
```

```python
# Create extracted PCA features

PCA_data = pd.DataFrame(
    X_pca,
    columns=["PC1", "PC2"]
)

PCA_data.head()
```

```python
# Check explained variance

print("Explained Variance Ratio:", pca.explained_variance_ratio_)
```

```python
# Apply PCA with all components

pca_full = PCA()

X_pca_full = pca_full.fit_transform(X_scaled)
```

```python
# Check explained variance ratio of all components

print("Explained Variance Ratio:")
print(pca_full.explained_variance_ratio_)
```

```python
# Calculate cumulative explained variance

cumulative_variance = pca_full.explained_variance_ratio_.cumsum()

print("Cumulative Explained Variance:")
print(cumulative_variance)
```

```python
# Feature Extraction using PCA - 7 Components

pca_final = PCA(n_components=7)

X_pca_final = pca_final.fit_transform(X_scaled)
```

```python
# Create PCA extracted features

PCA_features = pd.DataFrame(
    X_pca_final,
    columns=[
        "PC1",
        "PC2",
        "PC3",
        "PC4",
        "PC5",
        "PC6",
        "PC7"
    ]
)

PCA_features.head()
```

```python
# Check total explained variance

print(
    "Total Explained Variance:",
    pca_final.explained_variance_ratio_.sum()
)
```
